## Generations Table
The objective of this notebook is create a table with distinct generations extents on PokeAPI

```python
import requests
import pandas as pd
import duckdb
```

```python
# API url
url = "https://pokeapi.co/api/v2/generation/"

# Request data
response = requests.get(url)

# Verify request status
if response.status_code == 200:
    data = response.json()
    
    # Extract "results"
    generations = data.get("results", [])
    
    # Create DataFrame
    df_generations = pd.DataFrame(generations)
    df_generations = df_generations.drop(columns=['url'])
else:
    print(f"Error accessing API: {response.status_code}")
```

```python
# Add index and status columns
df_generations = df_generations.reset_index(drop=False)
df_generations['status'] = ""
```

```python
# Create DuckDB database
conn = duckdb.connect("/Users/andrebezerra/Desktop/Dev/data_with_duckdb/v2_database.duckdb")
conn.execute("""
    CREATE TABLE IF NOT EXISTS generations ( 
      INDEX INT,
      NAME TEXT,
      STATUS TEXT
    )
""")
conn.execute("""INSERT INTO generations SELECT * FROM df_generations""")
conn.close()
```
