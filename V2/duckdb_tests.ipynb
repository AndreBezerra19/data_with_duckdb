```python
import duckdb

# Conectar a um banco de dados DuckDB
conn = duckdb.connect('v2_database.duckdb', read_only=False)

# Executar o comando SHOW SCHEMAS para listar os esquemas existentes
result = conn.execute('SHOW TABLES').fetchall()

# Exibir os esquemas encontrados
print("Schemas existentes:")
for schema in result:
    print(schema[0])

# Fechar a conexão com o banco de dados DuckDB
conn.close()
```
